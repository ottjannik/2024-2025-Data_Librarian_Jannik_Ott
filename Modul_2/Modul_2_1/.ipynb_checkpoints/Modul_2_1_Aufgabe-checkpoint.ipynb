```python
url_base = "https://www.bibsonomy.org/json/search/"
url_attr = "?items=1000"
```

```python
def df_from_query(query):
    url = url_base + query.replace(" ", "%20") + url_attr  
    result = requests.get(url)
    json_data = result.json()
    dataframe = pd.DataFrame(json_data["items"])
    dataframe["query"] = query
    return dataframe
```

```python
dataframes = []
queries = ["database", "text", "library", "archive"]

for q in queries:
    df = df_from_query(q)
    dataframes.append(df)
    print(f"{len(df)} Ergebnisse für query {q} gefunden.")
```

```python
data = pd.concat(dataframes, ignore_index=True)
```

```python
df_filtered = data[['label', 'type', 'url', 'year', 'author', 'doi', 'query']]
```

```python
df_filtered = df_filtered.loc[data.type == "Publication"]
```

```python
df_filtered["year_correct"] = df_filtered.apply(lambda row: row.year.isdigit(), axis=1)
df_filtered = df_filtered.loc[df_filtered.year_correct == True]
df_filtered["year"].unique()
```

```python
delete_row = df_filtered.loc[df_filtered['year'] == '5555'].index
df_filtered.drop(delete_row, inplace=True, axis="index")
```

```python
df_filtered["year"].unique()
```

```python
df_filtered = df_filtered.loc[df_filtered["author"].isnull() == False]
df_filtered = df_filtered.drop_duplicates(subset = "label", keep ="first")
df_filtered = df_filtered.rename(columns={"label": "title", "author": "authors"})
```

```python
df_filtered = df_filtered.drop("year_correct", axis=1)
```
