```python
url = "https://www.bibsonomy.org/json/search/Bibliothek?items=1000&duplicates=merged"
```

```python
f = urllib.request.urlopen(url)
result = f.read().decode("utf-8")
```

```python
print(type(result))
```

```python
data = json.loads(result)
print(type(data))
```

```python
print(data)
```
