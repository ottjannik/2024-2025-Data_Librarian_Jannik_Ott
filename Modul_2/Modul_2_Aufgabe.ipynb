```python
import pandas as pd
import requests
```

```python
url = "https://www.bibsonomy.org/json/search/database?items=1000&duplicates=merged"
result = requests.get(url) # result ist ein requests.models.Response Objekt
data = result.json() # das Response-Objekt bietet diese nützliche Methode an
df_bib = pd.DataFrame(data["items"])
```
