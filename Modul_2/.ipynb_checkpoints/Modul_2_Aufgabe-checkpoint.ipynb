```python
df_selection = df[["label", "author", "year", "type", "count"]]
```

```python
df_selection = df_bib_selection.loc[df_bib_selection["author"].isnull() == False]
df_selection = df_bib_selection.loc[df_bib_selection["year"] <= "2025"]
df_selection = df_bib_selection.drop_duplicates(subset = "label", keep ="first")
```

```python
df_bib_selection["year"].unique()
```

```python
df_bib_selection["year_correct"] = df_bib_selection.apply(lambda row: row.year.isdigit(), axis=1)
```

```python
df_bib_selection = df_bib_selection.loc[df_bib_selection.year_correct == True]
df_bib_selection
```
