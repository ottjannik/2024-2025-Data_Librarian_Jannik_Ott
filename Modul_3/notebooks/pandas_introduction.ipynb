### 2.2 Skalenniveau und Datentypen

```python
import pandas as pd
data = {'month': ['Jan', 'Feb', 'Mar'],
        'temp': [-5, 2, 3],
         'below_zero': [True, False, False]}
df = pd.DataFrame(data)
print(df)
```

```python
# Welches Skalenniveau besitzen die Variablen im obigen Beispiel?
```

month: Nominalskala
temp: Metrische Skala (Intervallskala)
below_zero: Nominalskala

```python
# Sind die Variablen stetig oder diskret?
```

month: diskret
temp: stetig
below_zero: diskret
